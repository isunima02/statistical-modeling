```python
import pandas as pd
from scipy.stats import ttest_ind, mannwhitneyu, shapiro
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv("/content/TEH_World_Happiness_2019_Imputed.csv")

# Preview the data
df.head()
```

```python
# Remove any missing values
df_clean = df[['Country', 'GDP per capita', 'Happiness Score']].dropna()
```

```python
median_gdp = df_clean['GDP per capita'].median()
```

```python
# Create groups
high_gdp = df_clean[df_clean['GDP per capita'] > median_gdp]['Happiness Score']
low_gdp = df_clean[df_clean['GDP per capita'] <= median_gdp]['Happiness Score']

print(f"Median GDP per capita: {median_gdp:.4f}")
print(f"High GDP countries: {len(high_gdp)}")
print(f"Low GDP countries: {len(low_gdp)}")
```

```python
# Normality test
shapiro_high = shapiro(high_gdp)
shapiro_low = shapiro(low_gdp)

print(f"High GDP group normality p-value: {shapiro_high.pvalue:.4f}")
print(f"Low GDP group normality p-value: {shapiro_low.pvalue:.4f}")
```

```python
# Check if both groups are normally distributed
if shapiro_high.pvalue > 0.05 and shapiro_low.pvalue > 0.05:
    stat, p_value = ttest_ind(high_gdp, low_gdp, equal_var=False)
    test_used = "Independent t-test"
else:
    stat, p_value = mannwhitneyu(high_gdp, low_gdp, alternative='two-sided')
    test_used = "Mann–Whitney U test"
```

```python
print(f"\nTest Used: {test_used}")
print(f"Test Statistic: {stat:.4f}")
print(f"P-Value: {p_value:.8f}")
```

```python
# Add GDP group label to the original dataframe
df_clean['GDP Group'] = df_clean['GDP per capita'].apply(lambda x: 'High GDP' if x > median_gdp else 'Low GDP')
```

```python
# Plot
plt.figure(figsize=(8, 5))
sns.boxplot(x='GDP Group', y='Happiness Score', data=df_clean, palette="Set2")
plt.title("Happiness Score by GDP Group")
plt.ylabel("Happiness Score")
plt.show()
```
